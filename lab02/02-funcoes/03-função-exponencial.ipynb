### Função exponencial

#### <b style="color: #00008b">O que é?</b> 

Podemos dizer que a definição de função exponencial seria a seguinte:

<p>Dado um número real como a (sendo a > 0 e a &#8800; 1) denomina-se <i>função exponencial de base a</i>, a função <b style="color: #4F7942">f: <strong>&#8477;</strong> ->  <strong>&#8477;</strong> *+</b>: definida por <strong><i style="color: #D30000">f(x) = a<sup>x</sup></i></strong></p> 

<p>No trecho <q><b style="color: #4F7942">f: <strong>&#8477;</strong> ->  <strong>&#8477;</strong> *+</b></q> acima, é interessante destacar que o primeiro conjunto de números reais representados, que seria o ponto de partida, se refere ao domínio e o segundo, o de números reais positivos (de chegada), o contra-domínio (ler mais no artigo sobre <a href="https://github.com/octavioduarte/Matematica-Para-IA/blob/main/lab02/01-introducao-a-funcoes/conceitos.ipynb">funções</a>).</p>

#### <b style="color: #00008b">Exemplos</b>: 

<p><strong style="color: #FFDE21">a)</strong> <strong><i>f(x) = 2<sup>x</sup></i></strong> - &#x2705; <b>Sim, é uma função exponencial!</b> pois a variavel se encontra no expoente (elemento x), e a base é maior que 0 e diferente de 1.</p>

<p><strong style="color: #FFDE21">b)</strong> <strong><i>y = 3<sup>x</sup></i></strong> - &#x2705; <b>Sim, é uma função exponencial!</b> pois o valor da função (dessa vez representada como y, mas que pode ser também f(x), como no item anterior) possui 3 como base (regra do a > 0 e &#8800; 1) e expoente sendo a variável, com valor x.</p>

<p><strong style="color: #FFDE21">c)</strong> <strong><i>f(x) = ( <sup>1</sup>&frasl;<sub>3</sub> ) <sup>x</sup></i></strong> - &#x2705; <b>É uma função exponencial!</b> Uma vez que a base é uma fração que está em 0 e 1 (a > 0 e &#8800; 1), com expoente x.</p>

<p><strong style="color: #FFDE21">d)</strong> <strong><i>f(x) = (0,6) <sup>x</sup></i></strong> - &#x2705; <b>Também uma função exponencial!</b> Mesmo caso do item anterior, só que dessa vez, um valor decimal, mas que se enquadrada também na regra descrita anteriomente.</p>

<p><strong style="color: #FFDE21">e)</strong> <strong><i>f(x) = (-5) <sup>x</sup></i></strong> - &#x274C; <b>Não é uma função exponencial!</b> Apesar do expoente ser uma variável x, a base seria um número negativo que sendo menor que zero, faz com que não obedeça a regra de funções exponenciais.</p>

<p><strong style="color: #FFDE21">f)</strong> <strong><i>f(x) = 1 <sup>x</sup></i></strong> - &#x274C; <b>Não é uma função exponencial!</b> Pois claramente é possível notar que sendo a base o próprio valor 1, contraria a definição já citada.</p>

<p><strong style="color: #FFDE21">g)</strong> <strong><i>y = x <sup>2</sup></i></strong> - &#x274C; <b>Não é função exponencial!</b> Sendo a base a variável e o expoente o número dois, invertendo o que se entende por funções exponenciais, portanto, excluindo-se do conceito, em questão (se mostrando na realidade, uma função do segundo grau).</p>

<br/>

<p>Para praticarmos, fixando e explorando a resolução dos conceitos abordados, solucionaremos alguns exercícios, abaixo.</p>

<b style="color: #00008b">Dada a função exponencial <i>f(x)</i> = 4 <sup>x</sup> determine</b>: 

<p><strong style="color: #FFDE21">a)</strong> <strong><i>f(2)</i> <span style="color: #9F2B68"> = 4 <sup>2</sup> = 16</span><br/>
<span style="font-weight: 700; font-style: italic;">Observação: no exemplo acima, o número 16 é a IMAGEM do valor x que é igual a 2, na função do enunciado principal (<i>f(x)</i> = 4 <sup>x</sup>).</span>
</p>
    
<p><strong style="color: #FFDE21">b)</strong> <strong><i>f(-3)</i> <span style="color: #228B22"> = 4 <sup>-3</sup> = <sup>1</sup>&frasl;<sub>4<sup>3</sup></sub> = <sup>1</sup>&frasl;<sub>64</sub></span></p>

<p><strong style="color: #FFDE21">c)</strong> <strong><i>f(0,5)</i> <span style="color:#0096FF"> = 4 <sup>0,5</sup> = 4 <sup>1</sup>&frasl;<sub>2</sub></span> <span style="color:#FFBF00"> = <span style="white-space: nowrap">&radic;<span style="text-decoration:overline;">&nbsp;4&nbsp;</span></span></span> = 2</p>

<p><strong style="color: #FFDE21">d)</strong> <strong><i>m</i> tal que <i>f(m)</i> = 64,<span style="color: #CC5500"> ou seja,</span> <span style="color: #228B22"> 4 <sup>m</sup> = 64</span><span style="color:#0096FF"> => 4<sup>m</sup> = 4<sup>3</sup></span> <span style="color:#EE4B2B">(quando bases forem iguais, anulamos elas, deixando apenas os expoentes) => m = 3</span>.
<br/>
<span style="font-weight: 700; font-style: italic;">Observação: O último exemplo que vimos, localizado logo acima, se trata de uma EQUAÇÃO EXPONENCIAL.</span>
</p>
