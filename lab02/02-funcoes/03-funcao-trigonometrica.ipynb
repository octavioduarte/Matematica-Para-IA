<h2 style="color: red">Retas</h2>





Uma reta é um conjunto de pontos - Ela é representada por uma linha sem curva alguma.

É comum vermos em contextos matemáticos a frase: <i>"Dados dois pontos, existe <b>uma única reta</b> que os contém"</i> - Por tanto retas podem ser desenhadas a partir de apenas 2 pontos.

Retas possuem infinitos pontos <i>(acima vimos que podem ser traçadas a partir de 2, isso não significa que possuam apenas 2)</i> e se possuem infinitos pontos, significa que seu comprimento também é infinito.

<h3 style="color: red">Semirretas</h3>


Uma semirreta é um pedaço da reta, ela tem um começo, mas não tem fim.


<img src="./imagens/semirreta.webp" alt="Imagem ilustrando a representação de uma semireta" width="500"/>

Colocamos o ponto A que é onde a reta começa e o ponto B pois uma **reta deve ter no mínimo 2 pontos** - Entende-se que pelo ponto B passam diversas retas, mas somente **uma** passa pelo ponto A e B, por tanto devemos represetar uma reta com os 2 pontos.

<br/>

A representação desta semirreta fica desta forma:

<br/>

<img src="./imagens/representacao-de-uma-semirreta.png" alt="Representação de uma semireta" width="80"/>

Incluímos os pontos em qua reta começa e por onde ela passa com uma seta acima com pontas em uma das extremidades - O que indica que a reta é infinita para apenas um dos lados.

<h2 style="color: red">Ângulos</h2>

Ângulos são o resultado do encontro entre duas semirretas que tem a mesma origem. Temos 4 tipos de ângulos: agudo, reto, obtuso ou raso - Existem duas unidades de medidas para determinarmos o tamanho de um ângulo: <b>grau</b> e <b>radiano</b>.

É possível também estabelecer relações entre dois ângulos distintos:
 
 - Caso eles possuam a mesma medida - São chamados de <b>Congruentes</b>
 - Quando a soma entre eles é igual a 90º - São chamados de <b>Complementares</b>
 - Quando a soma entre eles é igual a 180º - São chamados de <b>Suplementares</b>
 - Quando a soma entre eles é igual a 360º - São chamados de <b>Replementares</b>

### Medição e unidades de medida de um ângulo

Já vimos que a unidade de medida para um ângulo é graus ou radianos - Vamos utilizar graus para seguir com os estudos,  A técnica de medir ângulos consiste em dividir um circulo em 360 partes iguais e cada parte equivare a 1 grau. 

No exemplo abaixo dividimos o circulo em 4 partes iguais e é possível visualizar os principais ângulos listados acima se conforme avançamos no sentido anti-horário

<img src="./imagens/círculo-trigonométrico-quadrantes.webp" alt="representação de um círculo-trigonométrico-quadrantes"/>

O 1° quadrante equivale a 90° graus

O 2° quadrante equivale a 180° graus

O 3° quadrante equivale a 270° graus

O 4° quadrante equivale a 360° graus
